```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.graphics import tsaplots
```

```python
train = pd.read_table('/Users/estevaoprado/Downloads/beijing+multi+site+air+quality+data/PRSA_Data_20130301-20170228/PRSA_Data_Gucheng_20130301-20170228.csv', sep=',')
```

```python
train.head()
y = train.dropna(subset=['PM2.5'])
y = np.asarray(y['PM2.5'])
print(y)
fig = tsaplots.plot_acf(y, lags=10)
plt.show()

fig = tsaplots.plot_pacf(y, lags=10)
plt.show()
```
